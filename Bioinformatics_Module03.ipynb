````python
from Bio import SeqIO
with open("rosalind_gc.txt","r") as data:
  line=data.read()
  highest_gc_id= ""
  highest_gc_content=-1.0
  for seq_record in SeqIO.parse ("rosalind_gc.txt", "fasta"):
    dna_sequence = str(seq_record.seq)
    gc_count = dna_sequence.count("G") + dna_sequence.count("C")
    gc_content = (gc_count / len(dna_sequence)) * 100

    if gc_content > highest_gc_content:
      highest_gc_content = gc_content
      highest_gc_id = seq_record.id
  print (highest_gc_id)
  print (highest_gc_content)
````
