```python
import pandas as pd

from sklearn.preprocessing import MultiLabelBinarizer

# Load your dataset
df = pd.read_csv("dummymatrix.csv")

# Strip leading and trailing whitespaces from column names to handle hidden spaces
df.columns = df.columns.str.strip()

# Identify isolate columns
isolate_columns = [
    "Isolate 1",
    "Isolate 2",
    "Isolate 3",
    "Isolate 4"
]

# Combine isolates from each sample into a list
df["Isolates"] = df[isolate_columns].apply(
    lambda row: [
        str(isolate).strip() for isolate in row
        if pd.notna(isolate) and isolate != 0 and str(isolate).strip() != "0" and str(isolate).strip() != ""
    ],
    axis=1
)

# One-hot encode the isolates
mlb = MultiLabelBinarizer()

encoded_isolates = pd.DataFrame(
    mlb.fit_transform(df["Isolates"]),
    columns=mlb.classes_,
    index=df.index
)

mlb = MultiLabelBinarizer()

encoded_isolates = pd.DataFrame(
    mlb.fit_transform(df["Isolates"]),
    columns=mlb.classes_,
    index=df.index
)

# Create predictor variables (X)
X = encoded_isolates

# Create target variable (y)
y = df["Host"]

print(X)
print(y)

```
